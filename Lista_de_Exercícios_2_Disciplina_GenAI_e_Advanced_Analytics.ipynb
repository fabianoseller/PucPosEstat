Prompt:

"Resolva de três formas distintas: Calcule o número total de parâmetros treináveis de uma camada Convolucional 2D com 32 filtros de tamanho 3x3, recebendo uma imagem RGB (3 canais), incluindo o bias."

```python
import torch
import torch.nn as nn

# Definição dos hiperparâmetros da camada
IN_CHANNELS = 3   # RGB
OUT_CHANNELS = 32 # Filtros
KERNEL_SIZE = 3   # 3x3
BIAS = True

# Instanciando a camada Convolucional
conv_layer = nn.Conv2d(in_channels=IN_CHANNELS,
                       out_channels=OUT_CHANNELS,
                       kernel_size=KERNEL_SIZE,
                       bias=BIAS)

print("--- DESAFIO DE AUTOCONSISTÊNCIA: CÁLCULO DE PARÂMETROS ---")

# --- MÉTODO 1: Lógica por Decomposição (Manual) ---
# Cada filtro tem (Kernel * Kernel * Canais de Entrada) pesos + 1 Bias
params_por_filtro = (KERNEL_SIZE * KERNEL_SIZE * IN_CHANNELS) + 1
total_metodo_1 = params_por_filtro * OUT_CHANNELS
print(f"Método 1 (Decomposição): {total_metodo_1} parâmetros")

# --- MÉTODO 2: Lógica por Shape de Tensores ---
# Pesos: [Out, In, H, W] | Bias: [Out]
weight_shape = conv_layer.weight.shape
bias_shape = conv_layer.bias.shape
total_metodo_2 = (weight_shape[0] * weight_shape[1] * weight_shape[2] * weight_shape[3]) + bias_shape[0]
print(f"Método 2 (Tensores): {total_metodo_2} parâmetros (Shape: {list(weight_shape)})")

# --- MÉTODO 3: Lógica Funcional (PyTorch built-in) ---
total_metodo_3 = sum(p.numel() for p in conv_layer.parameters() if p.requires_grad)
print(f"Método 3 (Built-in PyTorch): {total_metodo_3} parâmetros")

# --- VALIDAÇÃO FINAL ---
if total_metodo_1 == total_metodo_2 == total_metodo_3:
    print("\n[SUCESSO] O modelo é AUTOCONSISTENTE. Todos os caminhos levaram ao mesmo resultado.")
else:
    print("\n[ERRO] Houve divergência no raciocínio lógico.")
```

### Questão 5: Iteração (Otimização de Código)
Prompt 1: "Crie um loop simples em Python para normalizar uma lista de números entre 0 e 1."

Prompt 2: "Essa resposta está ruim. O código é ineficiente para grandes volumes de dados e não utiliza bibliotecas de performance que costumamos usar em Data Science."

Análise Crítica: O modelo deve evoluir de um for loop básico para uma solução usando NumPy ((x - min) / (max - min)). Avalie se ele "aprendeu" que você busca performance.

```python
import torch.optim as optim
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# 1. Geração de Dados Sintéticos (Regressão Linear)
torch.manual_seed(42)
np.random.seed(42)
X = torch.randn(100, 1) * 10
Y = 2.5 * X + 5 + torch.randn(100, 1) * 3  # y = 2.5x + 5 + ruído

# 2. Definição do Modelo e Função de Perda (Loss)
class LinearRegressionModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(1, 1)
    def forward(self, x):
        return self.linear(x)

criterion = nn.MSELoss()

# 3. Função de Treinamento Simulada (para o Gráfico Iterativo)
def treinar_modelo(learning_rate, epochs=100):
    model = LinearRegressionModel()
    # Começamos com pesos ruins para visualizar o aprendizado
    nn.init.constant_(model.linear.weight, 10.0)
    nn.init.constant_(model.linear.bias, 10.0)

    optimizer = optim.SGD(model.parameters(), lr=learning_rate)
    loss_history = []

    for epoch in range(epochs):
        # Forward pass
        outputs = model(X)
        loss = criterion(outputs, Y)

        # Backward e otimização
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        loss_history.append(loss.item())

    return loss_history, model

# 4. Visualização DEMONSTRATIVA E ITERATIVA (ipywidgets)
print("--- FERRAMENTA DE ITERAÇÃO DE PROMPT (SIMULAÇÃO) ---")
print("Ajuste o Slider abaixo para simular o feedback negativo.")
print("Um LR muito alto fará a Loss explodir. Um LR muito baixo fará o aprendizado ser lento.")

@interact(lr=FloatSlider(min=0.0001, max=0.01, step=0.0005, value=0.001, description='Learning Rate:'))
def plot_iterativo(lr):
    losses, model = treinar_modelo(lr)

    plt.figure(figsize=(12, 5))

    # Gráfico 1: Curva de Loss (Demonstrativo)
    plt.subplot(1, 2, 1)
    plt.plot(losses)
    plt.title(f'Curva de Loss (LR: {lr})')
    plt.xlabel('Época')
    plt.ylabel('MSE Loss')
    plt.yscale('log') # Escala logarítmica para ver explosão da loss
    plt.grid(True)

    # Gráfico 2: Ajuste da Reta (Iterativo)
    plt.subplot(1, 2, 2)
    plt.scatter(X.numpy(), Y.numpy(), alpha=0.5, label='Dados Reais')

    # Gera a reta predita
    X_range = torch.linspace(-30, 30, 100).view(-1, 1)
    Y_pred = model(X_range).detach().numpy()

    plt.plot(X_range.numpy(), Y_pred, color='red', linewidth=3, label='Predição do Modelo')
    plt.title('Ajuste da Reta Final')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()
```

# In-Context Learning na Prática: A capacidade do modelo de reagir à crítica é simulada pela função @interact. Quando você move o slider para um Learning Rate muito alto (ex: 0.01), a Loss explode (o gráfico da esquerda sobe). Seu feedback seria: "A resposta está ruim, o modelo não está convergindo!". Ao reduzir o LR (ex: 0.001), o modelo melhora a qualidade da predição.
