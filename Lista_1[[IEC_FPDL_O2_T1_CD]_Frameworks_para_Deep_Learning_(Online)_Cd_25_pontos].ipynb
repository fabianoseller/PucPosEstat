#**Lista de Exercicios 1**

O CIFAR-10 é um conjunto de dados amplamente utilizado em tarefas de aprendizado de máquina e visão computacional, especialmente para benchmarking de algoritmos de classificação de imagens. Aqui estão os principais detalhes sobre o CIFAR-10:
Descrição

- Nome Completo: CIFAR-10 (Canadian Institute For Advanced Research 10)
- Criadores: Alex Krizhevsky, Vinod Nair e Geoffrey Hinton.
- Ano de Publicação: 2009.
- Conteúdo: O CIFAR-10 contém 60.000 imagens coloridas de baixa resolução, organizadas em 10 classes diferentes. Cada imagem tem dimensões de 32x32 pixels e 3 canais de cor (RGB).

As 10 classes do CIFAR-10 são:

- Avião (Airplane)
- Automóvel (Automobile)
- Pássaro (Bird)
- Gato (Cat)
- Veado (Deer)
- Cachorro (Dog)
- Sapo (Frog)
- Cavalo (Horse)
- Navio (Ship)
- Caminhão (Truck)

Cada classe contém exatamente 6.000 imagens, resultando em um conjunto de dados equilibrado.
Divisão dos Dados

- Conjunto de Treinamento: 50.000 imagens.
- Conjunto de Teste: 10.000 imagens.

##**Orientações:** Execute a célula a seguir para fazer os devidos imports e em seguida complemente as células em branco.

```python
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (Conv2D,
                                     MaxPooling2D,
                                     Flatten,
                                     Dense)
from tensorflow.keras.utils import to_categorical

(X_train, y_train), (X_test, y_test) = cifar10.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0
y_train, y_test = to_categorical(y_train), to_categorical(y_test)

model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

## No código anterior, usamos validation_data=(X_test, y_test). Quando o professor usa validation_split=0.2, a lógica muda da seguinte forma:

###1. O que o validation_split=0.2 faz?
*Em vez de usar o conjunto de teste para validar, o Keras pega o seu conjunto de treino (X_train) e separa automaticamente os últimos 20% das imagens para servirem como validação.*
## Vantagem: Você mantém o conjunto de teste "puro", tocando nele apenas uma vez ao final de tudo (na Questão 3).
## Funcionamento: Se você tem 50.000 imagens de treino, 10.000 (20%) serão usadas para validar cada época, e o modelo treinará efetivamente com 40.000.

```python
# Questão 2: Treinar o modelo usando uma fração do treino para validação
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64,
    validation_split=0.2  # O Keras separa 20% do X_train para validar
)
```

```python
import matplotlib.pyplot as plt

# Plot the training and validation accuracy
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Acurácia de Treinamento')
plt.plot(history.history['val_accuracy'], label='Acurácia de Validação')
plt.title('Acurácia do Modelo por Época')
plt.xlabel('Época')
plt.ylabel('Acurácia')
plt.legend()

# Plot the training and validation loss
plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Perda de Treinamento')
plt.plot(history.history['val_loss'], label='Perda de Validação')
plt.title('Perda do Modelo por Época')
plt.xlabel('Época')
plt.ylabel('Perda')
plt.legend()

plt.tight_layout()
plt.show()
```

### **Questão 1**
Configure o modelo criado na célula anterior adicionando a função de perda, algoritmo otimizador e métrica


```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

###**Questão 2**
Treine o modelo anterior no dataset de treino definido no início do notebook, configurando os argumentos corretamente. Use 10 épocas para treinar o modelo.

```python
# Questão 2: Treinar o modelo
# Usaremos 10 épocas conforme sugerido no exercício
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64, # Tamanho do lote de imagens processadas por vez
    validation_data=(X_test, y_test)
)
```

###**Questão 3**
Avalie o modelo treinado anteriormente no dataset de teste e imprima a métrica

```python
# Questão 3: Avaliar o modelo no dataset de teste
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=2)

print(f'\nAcurácia no conjunto de teste: {test_acc*100:.2f}%')
```

# Dropout é uma das formas mais eficazes de combater o overfitting (quando o modelo fica "viciado" nos dados de treino e não consegue generalizar bem para imagens novas).

## O que o Dropout faz?
*Imagine que a sua rede neural é um time de basquete. Se um jogador for muito dominante, o resto do time fica "preguiçoso". O Dropout "desliga" aleatoriamente alguns neurônios durante o treinamento (ex: 20% deles). Isso força todos os neurônios a aprenderem características úteis sozinhos, tornando a rede muito mais robusta.*

### Para alterar o código
Você precisa importar a camada Dropout e inseri-la após as camadas com muitos parâmetros (geralmente após o MaxPooling ou a camada Dense).

## Modelo atualizado:

```python
from tensorflow.keras.layers import Dropout

model = Sequential([
    # Primeira parte: Convolução
    Conv2D(32, (3, 3), activation='relu', input_shape=(32, 32, 3)),
    MaxPooling2D((2, 2)),
    Dropout(0.25), # Desliga 25% dos neurônios aqui

    # Segunda parte: Convolução
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D((2, 2)),
    Dropout(0.25), # Desliga mais 25%

    # Parte de Classificação
    Flatten(),
    Dense(64, activation='relu'),
    Dropout(0.5),  # Camadas densas sofrem mais overfitting, usamos 50%
    Dense(10, activation='softmax')
])

# O restante (compile e fit) permanece igual!
```

```python
# Questão 2: Treinamento com separação automática de validação
history = model.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=64,
    # Aqui o Keras reserva 10.000 imagens do treino para validar
    # e treina efetivamente com as outras 40.000.
    validation_split=0.2
)
```

## Analisar os dados de teste e o desempenho do seu modelo no Google Colab, não basta apenas olhar para o número final da acurácia. É fundamental visualizar onde o modelo está acertando e em quais classes ele está confundindo (por exemplo, se ele acha que gato é cachorro).:

## Script para Matriz de Confusão e Relatório
Este código mostra exatamente quais classes o modelo mais confunde.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, classification_report

# 1. Obter as previsões do modelo para o conjunto de teste
y_pred = model.predict(X_test)
y_pred_classes = np.argmax(y_pred, axis=1)
y_true = np.argmax(y_test, axis=1)

# Nomes das classes do CIFAR-10
class_names = ['Avião', 'Automóvel', 'Pássaro', 'Gato', 'Veado',
               'Cachorro', 'Sapo', 'Cavalo', 'Navio', 'Caminhão']

# 2. Gerar a Matriz de Confusão
cm = confusion_matrix(y_true, y_pred_classes)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=class_names, yticklabels=class_names)
plt.xlabel('Predição do Modelo')
plt.ylabel('Classe Real')
plt.title('Matriz de Confusão - CIFAR-10')
plt.show()

# 3. Exibir métricas detalhadas (Precisão, Recall, F1-Score)
print(classification_report(y_true, y_pred_classes, target_names=class_names))
```

## Script para Visualizar Resultados na Prática
Nada melhor do que ver as imagens e a nota que o modelo deu para elas. O código abaixo plota 15 imagens de teste, indicando em verde se ele acertou e vermelho se errou.

```python
def plot_image_predictions(X, y_true, y_pred, classes, count=15):
    plt.figure(figsize=(16, 8))
    for i in range(count):
        plt.subplot(3, 5, i + 1)
        plt.imshow(X[i])

        true_label = classes[np.argmax(y_true[i])]
        pred_label = classes[y_pred[i]]

        color = 'green' if true_label == pred_label else 'red'

        plt.title(f"Real: {true_label}\nPred: {pred_label}", color=color)
        plt.axis('off')
    plt.tight_layout()
    plt.show()

# Chamando a função
plot_image_predictions(X_test, y_test, y_pred_classes, class_names)
```

### Por que analisar esses dados?Matriz de Confusão: Você notará que o modelo confunde muito "Gato" com "Cachorro" e "Caminhão" com "Automóvel".

### Isso acontece porque as características visuais (silhuetas, rodas, pelos) são muito parecidas para uma resolução de $32 \times 32$.Relatório (F1-Score):
*Se uma classe tiver um score muito baixo (ex: Pássaro), talvez você precise de mais imagens daquela categoria ou de uma rede mais profunda.Visualização: Às vezes o modelo erra, mas você percebe que até para um humano a imagem está confusa devido à baixa resolução.*

Script de Plotagem: Curvas de Perda e Acurácia

```python
import matplotlib.pyplot as plt

# Extraindo os dados do histórico de treinamento
acc = history.history['accuracy']
val_acc = history.history['val_accuracy']
loss = history.history['loss']
val_loss = history.history['val_loss']
epochs_range = range(1, len(acc) + 1)

plt.figure(figsize=(14, 5))

# Gráfico 1: Acurácia (Accuracy)
plt.subplot(1, 2, 1)
plt.plot(epochs_range, acc, label='Treino (Accuracy)')
plt.plot(epochs_range, val_acc, label='Validação (Val Accuracy)')
plt.title('Acurácia de Treino vs Validação')
plt.xlabel('Épocas')
plt.ylabel('Acurácia')
plt.legend(loc='lower right')
plt.grid(True)

# Gráfico 2: Perda (Loss)
plt.subplot(1, 2, 2)
plt.plot(epochs_range, loss, label='Treino (Loss)')
plt.plot(epochs_range, val_loss, label='Validação (Val Loss)')
plt.title('Perda de Treino vs Validação')
plt.xlabel('Épocas')
plt.ylabel('Erro (Loss)')
plt.legend(loc='upper right')
plt.grid(True)

plt.show()
```

# O Gap de Generalização: É o espaço entre a linha azul (treino) e a laranja (validação). Quanto maior o espaço, menos o seu modelo "generaliza" para dados novos.
## Sinal de Alerta (Overfitting):

No gráfico de Acurácia: Se a linha de treino sobe para 90% e a de validação trava em 70%, você tem overfitting.

## No gráfico de Loss (Perda): É o sinal mais claro. Se a linha de treino continua descendo, mas a de validação começa a subir, pare tudo! O modelo parou de aprender e começou a decorar.

# CIFAR-10?
As imagens são pequenas e o dataset é complexo. Sem camadas como Dropout ou Batch Normalization, a rede neural rapidamente "decora" as 50.000 imagens de treino.

Análise das Curvas de Aprendizado
O gráfico que você gerou fornece sinais claros sobre a saúde do seu modelo após as 10 épocas solicitadas:

Gap de Generalização: Há um distanciamento constante entre as curvas de treino e validação. No seu gráfico de acurácia, a curva de Validação (Laranja) está, curiosamente, acima da de Treino (Azul). Isso pode ocorrer no início do treinamento devido ao uso de camadas de regularização (como Dropout) que são desativadas na validação, ou simplesmente porque o conjunto de validação foi "mais fácil" para o modelo neste estágio inicial.

Identificando o Overfitting: No seu gráfico atual, ainda não há um overfitting severo. O overfitting clássico é identificado quando a Perda de Treino continua caindo, mas a Perda de Validação começa a subir. No seu gráfico de Perda (Loss), ambas continuam em trajetória de queda, o que sugere que o modelo ainda tinha capacidade de aprender se fosse treinado por mais épocas.

Convergência: As curvas estão inclinadas, indicando que o modelo não "estacionou". Com apenas 10 épocas, você capturou apenas o início do aprendizado.

# Script de Diagnóstico para o Colab
Para complementar sua análise e documentar esse "gap" formalmente no seu notebook, você pode utilizar o seguinte bloco de código:

```python
# Cálculo do Gap Final
final_train_acc = history.history['accuracy'][-1]
final_val_acc = history.history['val_accuracy'][-1]
gap = abs(final_train_acc - final_val_acc)

print(f"Acurácia de Treino Final: {final_train_acc:.4f}")
print(f"Acurácia de Validação Final: {final_val_acc:.4f}")
print(f"Gap de Generalização: {gap:.4f}")

if final_train_acc > final_val_acc + 0.1:
    print("Diagnóstico: Alerta de Overfitting detectado.")
else:
    print("Diagnóstico: Modelo estável, mas pode melhorar com mais épocas.")
```
