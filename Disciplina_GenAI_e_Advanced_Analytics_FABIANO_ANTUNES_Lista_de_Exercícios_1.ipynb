```python
# Instalação das bibliotecas necessárias
!pip install langchain langchain-openai langchain-core -q

import os
from google.colab import userdata
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# Configuração da API Key (Certifique-se de ter a OPENAI_API_KEY nos segredos do Colab)
os.environ['OPENAI_API_KEY'] = userdata.get('OPENAI_KEY')

# Inicialização do LLM conforme padrão do seu projeto [cite: 43, 44, 45]
llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0.7
)
```

```python
# Dicionário com as questões da lista
questoes = {
    "Questão 1": "Explique o conceito de Self-Attention nos Transformers comparando-o com uma situação do cotidiano, como uma festa ou uma conversa em grupo.",
    "Questão 2": "Por que modelos baseados apenas em Decoder, como o GPT, tornaram-se mais populares para o público geral do que modelos baseados em Encoder (como o BERT)?",
    "Questão 3": "Cite exemplos de 'capacidades emergentes' que surgiram em modelos maiores (como o GPT-4) que não eram vistas em modelos menores.",
    "Questão 4": "Pode me dar detalhes sobre a Guerra Civil em Plutão no ano de 2015?",
    "Questão 5": "Sugira 3 casos de uso de GenAI para uma empresa de capital aberto no setor financeiro."
}

# Loop de execução
resultados = {}

print("Iniciando a geração das respostas...\n")

for titulo, prompt_texto in questoes.items():
    chat_prompt = ChatPromptTemplate.from_messages([
        ("system", "Você é um assistente acadêmico especializado em IA e Analytics."),
        ("human", "{input}")
    ])

    chain = chat_prompt | llm
    response = chain.invoke({"input": prompt_texto})

    resultados[titulo] = response.content
    print(f"✅ {titulo} finalizada.")

print("\n--- PROCESSAMENTO CONCLUÍDO ---")
```

```python
for titulo, resposta in resultados.items():
    print(f"\n{'='*30}")
    print(f"ESTRUTURA PARA O RELATÓRIO: {titulo}")
    print(f"{'='*30}")
    print(f"PROMPT UTILIZADO: {questoes[titulo]}")
    print(f"\nRESUMO DA RESPOSTA:\n{resposta[:500]}... (veja resposta completa acima)")
    print(f"\n[ESPAÇO PARA SUA ANÁLISE CRÍTICA]")
```

```python
from docx import Document
from docx.shared import Pt, Inches
from docx.enum.text import WD_ALIGN_PARAGRAPH
from google.colab import files

def adicionar_tabela_reflexion(doc, inicial, critica, melhorada):
    doc.add_heading('Comparativo de Refinamento (Reflexion)', level=2)
    table = doc.add_table(rows=1, cols=3)
    table.style = 'Light Grid Accent 1'

    # Cabeçalho da Tabela
    hdr_cells = table.rows[0].cells
    hdr_cells[0].text = 'Resposta Inicial (Analista)'
    hdr_cells[1].text = 'Crítica (Red Team / Risco)'
    hdr_cells[2].text = 'Resposta Melhorada (CIO)'

    # Linha de Conteúdo
    row_cells = table.add_row().cells
    row_cells[0].text = inicial
    row_cells[1].text = critica
    row_cells[2].text = melhorada

def gerar_docx_completo(resultados, questoes, dados_reflexion=None):
    doc = Document()
    doc.add_heading('Lista de Exercícios 1: GenAI e Advanced Analytics', 0)

    for titulo, prompt_texto in questoes.items():
        doc.add_heading(titulo, level=1)

        # Seção de Prompt e Resumo Geral
        doc.add_heading('Prompt Utilizado:', level=2)
        doc.add_paragraph(prompt_texto)

        doc.add_heading('Resumo da Resposta:', level=2)

        # Se for a Questão 5 e tivermos dados de Reflexion, geramos a tabela
        if titulo == "Questão 5" and dados_reflexion:
            adicionar_tabela_reflexion(
                doc,
                dados_reflexion.get('initial_answer', ''),
                dados_reflexion.get('critique', ''),
                dados_reflexion.get('improved_answer', '')
            )
        else:
            doc.add_paragraph(resultados[titulo])

        # Espaço para sua Análise Crítica (Obrigatório)
        doc.add_heading('Sua Análise Crítica (Curadoria):', level=2)
        p = doc.add_paragraph()
        p.add_run('[Escreva aqui sua avaliação técnica: o modelo alucinou? O tom foi adequado? A complexidade foi bem tratada?]').italic = True

        doc.add_page_break()

    nome_arquivo = 'Entrega_GenAI_Final.docx'
    doc.save(nome_arquivo)
    files.download(nome_arquivo)

# Exemplo de como chamar com os dados que você já gerou no seu notebook
# reflexion_result deve conter as chaves: 'initial_answer', 'critique', 'improved_answer'

# Define a dummy reflexion_result for demonstration purposes
reflexion_result = {
    'initial_answer': resultados.get('Questão 5', 'Resposta inicial da Questão 5 não disponível.'),
    'critique': 'Crítica: A resposta poderia ser mais detalhada em alguns pontos e considerar mais riscos específicos do setor financeiro.',
    'improved_answer': 'Resposta melhorada: Implementar IA para análise de risco em tempo real, detecção de fraudes avançada e personalização de serviços financeiros com maior foco em compliance.'
}

gerar_docx_completo(resultados, questoes, dados_reflexion=reflexion_result)
```
